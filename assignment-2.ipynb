```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

digits = load_digits()
print("Dataset Shape:", digits.data.shape)
print("Image Shape:", digits.images.shape)
print("Target Levels:", digits.target_names)

plt.figure(figsize=(4,4))
plt.imshow(digits.images[0], cmap='pink')
plt.title(f"Label:{digits.target[99]}")
plt.axis('off')
plt.show()
```

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

digits = load_digits()
fig, axes = plt.subplots(2, 5, figsize=(10, 5))

for i, ax in enumerate(axes.flat):
    ax.imshow(digits.images[i], cmap='gray') 
    ax.set_title(f"Digit: {digits.target[i]}")
    ax.axis('off')
plt.tight_layout()
plt.show()   
```

**2.Write a program to display:-
(a)number of rows and columns
(b)column names 
(c)data types**


```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['target'] = iris.target

print("Shape (rows, columns):", df.shape)

print("\nColumn names:", list(df.columns))

print("\nData types:\n", df.dtypes)
```

**3.Write a program to count the number of samples in iris dataset and digit dataset.**

```python
from sklearn.datasets import load_iris, load_digits

iris = load_iris()
digits = load_digits()

print("Number of samples in Iris dataset:", iris.data.shape[0])
print("Number of samples in Digits dataset:", digits.data.shape[0])
```
